```python
#Source location for the MIMIC-III demo data
SOURCE_PATH = "s3://mimic-iii-demo-bucket/"

#List everything currently available in the bucket
files = dbutils.fs.ls(SOURCE_PATH)

file_inventory = spark.createDataFrame(
    [(f.name, f.size, f.path) for f in files],
    ["file_name", "size_bytes", "path"]
)

display(
    file_inventory
    .orderBy("file_name")
)
```

```python
def load_mimic_csv(file_name):
    return (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv(f"{SOURCE_PATH}{file_name}")
    )


admissions = load_mimic_csv("ADMISSIONS.csv")
patients = load_mimic_csv("PATIENTS.csv")
icustays = load_mimic_csv("ICUSTAYS.csv")

print(f"ADMISSIONS: {admissions.count():,} rows, {len(admissions.columns)} columns")
print(f"PATIENTS:   {patients.count():,} rows, {len(patients.columns)} columns")
print(f"ICUSTAYS:   {icustays.count():,} rows, {len(icustays.columns)} columns")

display(admissions.limit(10))
```

```python
print("ADMISSIONS SCHEMA")
admissions.printSchema()

print("\nPATIENTS SCHEMA")
patients.printSchema()

print("\nICUSTAYS SCHEMA")
icustays.printSchema()
```

```python
print("PATIENTS")
print(f"Rows: {patients.count():,}")
print(f"Unique subject_id: {patients.select('subject_id').distinct().count():,}")

print("\nADMISSIONS")
print(f"Rows: {admissions.count():,}")
print(f"Unique subject_id: {admissions.select('subject_id').distinct().count():,}")
print(f"Unique hadm_id: {admissions.select('hadm_id').distinct().count():,}")

print("\nICUSTAYS")
print(f"Rows: {icustays.count():,}")
print(f"Unique subject_id: {icustays.select('subject_id').distinct().count():,}")
print(f"Unique hadm_id: {icustays.select('hadm_id').distinct().count():,}")
print(f"Unique icustay_id: {icustays.select('icustay_id').distinct().count():,}")
```
