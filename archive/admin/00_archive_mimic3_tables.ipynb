# Archive MIMIC-III Tables

This notebook archives the previous MIMIC-III Bronze and Silver tables before development moves to the MIMIC-IV dataset.

```python
# Create archive schemas for the previous MIMIC-III tables
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.archive_mimic3_bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.archive_mimic3_silver")

print("Archive schemas created")
```

## Check Existing MIMIC-III Tables

Confirm the existing Bronze and Silver tables before moving them into the archive schemas.

```python
print("BRONZE TABLES")
display(spark.sql("SHOW TABLES IN workspace.bronze"))

print("SILVER TABLES")
display(spark.sql("SHOW TABLES IN workspace.silver"))
```

## Archive MIMIC-III Tables

Move the previous MIMIC-III Bronze and Silver tables into dedicated archive schemas before starting the MIMIC-IV pipeline.

```python
# Move MIMIC-III Bronze tables
bronze_tables = [
    "callout_raw",
    "caregivers_raw",
    "chartevents_raw",
    "d_items_raw",
    "datetimeevents_raw"
]

for table in bronze_tables:
    spark.sql(
        f"ALTER TABLE workspace.bronze.{table} "
        f"RENAME TO workspace.archive_mimic3_bronze.{table}"
    )

# Move MIMIC-III Silver tables
silver_tables = [
    "callout",
    "caregivers",
    "chartevents",
    "d_items",
    "datetimeevents"
]

for table in silver_tables:
    spark.sql(
        f"ALTER TABLE workspace.silver.{table} "
        f"RENAME TO workspace.archive_mimic3_silver.{table}"
    )

print("MIMIC-III tables archived")
```
