# Patient Journey - Gold Layer

This notebook creates an admission-level Gold dataset for hospital operations and patient journey analysis.

The final table contains one row per hospital admission (`hadm_id`) and combines patient demographics, admission details, hospital transfers, clinical services, ICU utilisation, diagnoses and procedures.

The dataset is designed for Databricks dashboards, Genie analytics and role-based security demonstrations.

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

CATALOG = "workspace"
SILVER_SCHEMA = "silver"
GOLD_SCHEMA = "gold"

TARGET_TABLE = f"{CATALOG}.{GOLD_SCHEMA}.patient_journey"

required_tables = [
    "patients",
    "admissions",
    "transfers",
    "services",
    "icustays",
    "diagnoses_icd",
    "d_icd_diagnoses",
    "procedures_icd",
    "d_icd_procedures"
]

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{GOLD_SCHEMA}")

missing_tables = []

for table in required_tables:
    full_name = f"{CATALOG}.{SILVER_SCHEMA}.{table}"

    if spark.catalog.tableExists(full_name):
        print(f"READY   {full_name}")
    else:
        print(f"MISSING {full_name}")
        missing_tables.append(table)

print()
print(f"Required tables: {len(required_tables)}")
print(f"Available tables: {len(required_tables) - len(missing_tables)}")
print(f"Missing tables: {len(missing_tables)}")
```
